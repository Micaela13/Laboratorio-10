## 1. Preguntas de Gestión Seleccionadas

Para este tablero, diseñamos la interfaz enfocada en responder interrogantes estratégicas para directivos del sector turístico, reutilizando los hallazgos analíticos previos:

1. **¿Cómo ha evolucionado el flujo de visitantes y qué impacto real dejó la pandemia y su posterior recuperación?**
   * *Utilidad directiva:* Permite dimensionar la caída histórica (~73% en 2020) y la estabilización post-pandemia para proyectar la infraestructura necesaria.
2. **¿Cuáles son los mercados emisores y vías de ingreso críticas en las que debemos enfocar la promoción?**
   * *Utilidad directiva:* Identifica que El Salvador, Estados Unidos y Guatemala concentran el mayor volumen, predominando la vía terrestre y el aeropuerto de La Aurora.
3. **¿Qué tan confiables son las predicciones de nuestros modelos (Prophet, Holt-Winters, ARIMA) para la planificación?**
   * *Utilidad directiva:* Evalúa el error (MAE y RMSE) de diferentes enfoques predictivos según el comportamiento estacional o de quiebre estructural de cada serie.

## 4. Guía de Diseño, Paleta de Colores y UX/HCI

### 4.1 Elección de Colores e Identidad Visual
Inspirados en una estética orgánica, natural y de alta legibilidad para entornos directivos:
* **Fondo Neutro / Crema (`#F3F4F0`):** Fondo general para evitar la fatiga visual.
* **Verde Bosque / Principal (`#2D5A27`):** Utilizado en menús, barras principales y elementos clave. Transmite estabilidad y sostenibilidad.
* **Azul Petróleo / Secundario (`#1E40AF`):** Usado para gráficos de tendencias temporales y elementos interactivos de soporte.
* **Tonos Arena y Crema Oscuro (`#D1D5DB` / `#E5E7EB`):** Para separar contenedores, tarjetas y fondos de gráficos secundarios.
* **Terracota / Alerta (`#C2410C`):** Color de acento reservado exclusivamente para resaltar puntos de sobreocupación turística, alertas de capacidad o anomalías de error en los modelos.

### 4.2 Criterios UX/HCI Aplicados
* Interfaz orientada a la simplicidad para usuarios no estadísticos, separando con claridad los hallazgos observados de las estimaciones analíticas.
* Consistencia semántica estricta en el uso de los colores en todas las pestañas del tablero.
